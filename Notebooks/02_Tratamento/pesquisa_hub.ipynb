```python
# Função vetorizada para cálculo de distâncias
def calcular_distancias_vetorizado(lat_mun, lon_mun, df_imoveis):
    """
    Calcula distâncias usando fórmula de Haversine vetorizada
    """
    lat1 = np.radians(lat_mun)
    lon1 = np.radians(lon_mun)
    lat2 = np.radians(df_imoveis['Latitude'].values)
    lon2 = np.radians(df_imoveis['Longitude'].values)
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    
    return 6371 * c  # Raio da Terra em km
```

```python
# Função para analisar imóveis por faixa etária
def analisar_imoveis_por_faixa_etaria(df_imoveis, df_idade, faixas_etarias, 
                                        top_n=10, raio_km=50):
    """
    Analisa imóveis próximos aos municípios com maior população em cada faixa etária
    
    Parâmetros:
    - df_imoveis: DataFrame com os imóveis
    - df_idade: DataFrame com dados demográficos dos municípios
    - faixas_etarias: Lista de faixas etárias para analisar
    - top_n: Número de municípios por faixa
    - raio_km: Raio de busca em km
    """
    
    print(f"\nProcessando {len(faixas_etarias)} Faixas Etárias")
    
    resultados_por_faixa = {}
    df_imoveis_valid = df_imoveis.dropna(subset=['Latitude', 'Longitude']).copy()
    
    # Processar cada faixa etária
    for faixa_etaria in faixas_etarias:
        print(f"\n→ {faixa_etaria}...", end=' ')
        
        # Validar colunas necessárias
        cols_req = ['latitude', 'longitude', 'municipio', 'sigla_uf']
        if not all(c in df_idade.columns for c in cols_req):
            print(" Colunas faltando")
            continue
        
        # Preparar dados dos municípios
        df_mun = df_idade.copy()
        df_mun[['latitude', 'longitude']] = df_mun[['latitude', 'longitude']].apply(
            pd.to_numeric, errors='coerce'
        )
        df_mun = df_mun.dropna(subset=['latitude', 'longitude'])
        
        # Verificar se a faixa etária existe
        faixa_col = faixa_etaria.lower()
        if faixa_col not in df_mun.columns:
            print(" Não encontrada")
            continue
        
        # Selecionar top municípios
        df_mun[faixa_col] = pd.to_numeric(df_mun[faixa_col], errors='coerce')
        df_top = df_mun.nlargest(top_n, faixa_col)
        
        resultados = []
        
        # Processar cada município
        for _, mun in df_top.iterrows():
            distancias = calcular_distancias_vetorizado(
                mun['latitude'], mun['longitude'], df_imoveis_valid
            )
            
            mask = distancias <= raio_km
            proximos = df_imoveis_valid[mask].copy()
            
            if len(proximos) > 0:
                proximos['distancia_km'] = distancias[mask]
                proximos['municipio_ref'] = mun['municipio']
                proximos['uf_ref'] = mun['sigla_uf']
                proximos['pop_faixa'] = mun[faixa_col]
                proximos['lat_mun'] = mun['latitude']
                proximos['lon_mun'] = mun['longitude']
                resultados.append(proximos)
        
        if resultados:
            df_resultado = pd.concat(resultados, ignore_index=True)
            df_resultado = df_resultado.sort_values('distancia_km').drop_duplicates(
                subset=['Latitude', 'Longitude'], keep='first'
            )
            resultados_por_faixa[faixa_etaria] = df_resultado
            print(f"1 {len(df_resultado)} imóveis")
        else:
            print(" Nenhum imóvel")
    
    if not resultados_por_faixa:
        print("\n Nenhum resultado")
        return None
    
    return resultados_por_faixa
```

```python
def criar_mapa_interativo(resultados_por_faixa):
    """
    Cria mapa interativo com botões para alternar entre grupos etários
    """
    # Agrupar faixas etárias
    grupos = {
        'Jovem (15-29)': ['15 a 19 anos', '20 a 29 anos'],
        'Adulto (30-59)': ['30 a 39 anos', '40 a 49 anos', '50 a 59 anos'],
        'Idoso (60-79)': ['60 a 69 anos', '70 a 79 anos'],
        'Idoso 80+': ['80 anos e mais']
    }
    
    # Consolidar resultados por grupo
    resultados_agrupados = {}
    for grupo, faixas in grupos.items():
        dfs_grupo = []
        for faixa in faixas:
            if faixa in resultados_por_faixa:
                dfs_grupo.append(resultados_por_faixa[faixa])
        
        if dfs_grupo:
            resultados_agrupados[grupo] = pd.concat(dfs_grupo, ignore_index=True)
    
    # Calcular totais para percentuais
    total_imoveis = sum(len(df) for df in resultados_agrupados.values())
    
    fig = go.Figure()
    
    # Cores para cada grupo
    cores = {
        'Jovem (15-29)': '#FF6B6B',
        'Adulto (30-59)': '#4ECDC4',
        'Idoso (60-79)': '#FFA07A',
        'Idoso 80+': '#BB8FCE'
    }
    
    # Adicionar traços para cada grupo
    for i, (grupo, df_resultado) in enumerate(resultados_agrupados.items()):
        percentual = (len(df_resultado) / total_imoveis) * 100
        cor = cores.get(grupo, '#4ECDC4')
        
        fig.add_trace(go.Scattermap(
            lat=df_resultado['Latitude'], 
            lon=df_resultado['Longitude'],
            mode='markers',
            marker=dict(
                size=15,
                color=cor,
                opacity=0.8
            ),
            name=grupo,
            customdata=np.column_stack([
                df_resultado['listing.address.city'],
                df_resultado['listing.address.neighborhood'],
                df_resultado['listing.pricingInfo.price'],
                df_resultado['listing.usableAreas'],
                df_resultado['listing.bedrooms'],
                df_resultado['distancia_km'],
                df_resultado['municipio_ref'],
                [f"{percentual:.1f}%" for _ in range(len(df_resultado))]
            ]),
            hovertemplate=
                '<b>%{customdata[0]} - %{customdata[1]}</b><br>' +
                'Preço: R$ %{customdata[2]:,.2f}<br>' +
                'Área: %{customdata[3]:.0f}m² | Quartos: %{customdata[4]:.0f}<br>' +
                'Distância: %{customdata[5]:.1f}km → %{customdata[6]}<br>' +
                '<b>Grupo: ' + grupo + ' (%{customdata[7]})</b><extra></extra>',
            visible=True
        ))
    
    # Criar botões para alternar grupos
    buttons = []
    for i, (grupo, df_resultado) in enumerate(resultados_agrupados.items()):
        visible = [False] * len(fig.data)
        visible[i] = True
        percentual = (len(df_resultado) / total_imoveis) * 100
        
        buttons.append(dict(
            label=f"{grupo} ({percentual:.1f}%)",
            method='update',
            args=[
                {'visible': visible},
                {'title': f'{grupo} | {len(df_resultado)} imóveis ({percentual:.1f}% do total)'}
            ]
        ))
    
    # Configurar layout inicial
    primeiro_grupo = list(resultados_agrupados.keys())[0]
    df_primeiro = resultados_agrupados[primeiro_grupo]
    percentual_primeiro = (len(df_primeiro) / total_imoveis) * 100
    
    fig.update_layout(
        mapbox=dict(
            style='open-street-map',
            center=dict(
                lat=df_primeiro['Latitude'].mean(),
                lon=df_primeiro['Longitude'].mean()
            ),
            zoom=6
        ),
        title=dict(
            text=f"{primeiro_grupo} | {len(df_primeiro)} imóveis ({percentual_primeiro:.1f}% do total)",
            font=dict(size=18)
        ),
        updatemenus=[
            dict(
                buttons=buttons,
                direction='down',
                pad={'r': 10, 't': 10},
                showactive=True,
                x=0.01,
                xanchor='left',
                y=0.98,
                yanchor='top',
                bgcolor='rgba(255,255,255,0.95)',
                bordercolor='#333',
                borderwidth=2,
                font=dict(size=11)
            )
        ],
        showlegend=False,
        height=800,
        margin=dict(r=0, t=80, l=0, b=0)
    )
    
    return fig
```
