```python
#Carregamento das bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from unidecode import unidecode
from math import radians, cos, sin, asin, sqrt
import plotly.graph_objects as go
```

```python
# Carregamento do dataset
df = pd.read_csv('../../Data/Raw/imovel_br.csv', sep=';', encoding='utf-8', low_memory=False)
```

```python
#Pré-processamento dos dados
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str)                    # garante que é string
    df[col] = df[col].apply(lambda x: unidecode(x))  # remove acentos
    df[col] = df[col].str.upper()  
```

```python
#Análise inicial dos dados
df.info()
df.head()
```

```python
# Listagem das colunas do DataFrame
print(df.columns.tolist())
```

```python
# Renomear colunas para facilitar o acesso
df = df.rename(columns={
    'listing.address.point.lat': 'Latitude',
    'listing.address.point.lon': 'Longitude'
})
```

```python
# Remover linhas com valores ausentes em Latitude e Longitude
df = df.dropna(subset=["Latitude", "Longitude"])
```

```python
# Converter colunas de Latitude e Longitude para numérico
df['Latitude'] = pd.to_numeric(df['Latitude'], errors='coerce')
df['Longitude'] = pd.to_numeric(df['Longitude'], errors='coerce')
```

```python
# Remover linhas com valores ausentes em Latitude e Longitude após conversão
df = df.dropna(subset=['Latitude', 'Longitude'])
```

```python
# Verificar os tipos de dados das colunas relevantes
df[["Latitude", "Longitude", "listing.totalAreas"]].info()
```

```python
# Criar uma coluna de tipo de imóvel mais descritiva
def get_property_type(row):
    # Tentar pegar de unitTypes primeiro
    if pd.notna(row.get('listing.unitTypes')) and row.get('listing.unitTypes'):
        return str(row['listing.unitTypes'])
    # Senão, tentar unitSubTypes
    elif pd.notna(row.get('listing.unitSubTypes')) and row.get('listing.unitSubTypes'):
        return str(row['listing.unitSubTypes'])
    # Senão, tentar usageTypes
    elif pd.notna(row.get('listing.usageTypes')) and row.get('listing.usageTypes'):
        return str(row['listing.usageTypes'])
    # Senão, usar o tipo genérico
    else:
        return str(row.get('listing.propertyType', 'Não especificado'))

df['Tipo de Imóvel'] = df.apply(get_property_type, axis=1)
```

```python
# Verificar o conteúdo das colunas relacionadas a tipo
print("Verificando colunas de tipo\n")

print("listing.propertyType:")
print(df['listing.propertyType'].value_counts())
print("\n" + "="*30 + "\n")

print("listing.unitTypes:")
print(df['listing.unitTypes'].value_counts())
print("\n" + "="*30 + "\n")

print("listing.unitSubTypes:")
print(df['listing.unitSubTypes'].value_counts())
print("\n" + "="*30 + "\n")

print("listing.usageTypes:")
print(df['listing.usageTypes'].value_counts())
print("\n" + "="*30 + "\n")
print("imvl_type:")
print(df['imvl_type'].value_counts())
```

```python
# Visualização dos dados geográficos usando Plotly Express
fig = px.scatter_map(
    df,
    lat="Latitude",
    lon="Longitude",
    color="listing.address.city",
    hover_name="listing.address.neighborhood",
    hover_data={
        "listing.unitTypes": True,
        "listing.pricingInfo.price": True,
        "listing.usableAreas": True,
        "listing.bedrooms": True,
        "listing.bathrooms": True,
        "listing.parkingSpaces": True,
        "listing.address.city": True,
        "listing.address.state": True,
        "listing.pricingInfo.monthlyCondoFee": True,
        "listing.pricingInfo.yearlyIptu": True,
        "Latitude": False,
        "Longitude": False
    },
    zoom=4,
    height=600,
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_layout(
    showlegend=False,
    title={
        'text': "Distribuição de Imóveis nas Principais Cidades Brasileiras",
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 18}
    },
    margin={"r": 0, "t": 60, "l": 0, "b": 0}
)

fig.update_traces(
    marker=dict(size=8, opacity=0.7)
)

fig.show()
```
